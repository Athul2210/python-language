```python
#2. Checking methods - Returns boolean
#  isalpha() - Checks whether the text contains only letters.
print("Hello".isalpha())
print("hello1".isalpha())

#isdigit() - Checks whether the text contains only digits.
print("Hello".isdigit())
print("2764".isdigit())

#isalnum() - Checks whether the text contains only letters and numbers. It doesn't allow spaces or special symbols.
print("HEllo123".isalnum())

#islower()- Checks whether all the letters are lowercase.
print("hello".islower())

#isupper() - Checks whether all the letters are uppercase.
print("HELLO".isupper())

#istitle() - Checks whether each word starts with a capital letter.
print("Hello World".istitle())

#isspace() - Checks whether the text contains only spaces or other whitespace characters.
print("  ".isspace())
print("Hello".isspace())

#startswith() - Checks whether a sentence starts with a particular word or letter.
print("Python is easy".startswith("Python"))

#endswith() - Checks whether a sentence ends with a particular word or letter.
print("Python is easy".endswith("easy"))
```

```python
# 3. Removing spaces - Imagine you have extra spaces before and after a word. These methods help remove them.
# strip() - Removes spaces from both sides.
s="   Python   "
print(s.strip())

#lstrip() - Removes spaces from the left side only.
text = "   Hello   World"
print(text.lstrip())

#rstrip - Removes spaces from the right side only.
text = "      Hello World     "
print(text.rstrip())

#Removing a particular character - You can also remove a particular character by passing it inside the brackets.
text = "###Hello###"
print(text.strip("#"))

#4. Replacing text
#  replace() - Replaces one word or letter with another.
text = "My name is Adithya"
print(text.replace("Adithya", "Adarsh"))

#maketrans() and translate() - These are used together to replace multiple characters.
text = "cat"
table = str.maketrans("cat", "123")
print(text.translate(table))
```

```python
# 5. Splitting and joining - These methods are used to break sentences into words or combine words into a sentence.
#split() - Breaks a sentence into separate words and puts them in a list.
text = "I love Python"
print(text.split())

#rsplit() - Splits a sentence from the right side. You can tell it how many times to split.
text = "apple-banana-orange"
print(text.rsplit("-", 1))     #It splits only once, starting from the right.

#splitlines()- Splits text wherever there is a new line.
text = "Hello\nPython"
print(text.splitlines())

#join() - Joins separate words together using a separator.
words = ["I", "love", "Python"]
print(" ".join(words))

#partition()- Divides a sentence into three parts at the first occurrence of a character.
text = "apple-banana-orange"
print(text.partition("-"))

#rpartition()- Divides a sentence into three parts at the last occurrence of a character.
text = "apple-banana-orange"
print(text.rpartition("-"))
```

```python
#6. String alignment methods- These methods help you arrange text in a particular space.
# center()- Places text in the center of a given width.
print("Hi".center(10, "*"))

#ljust()- Places text on the left and fills the remaining space on the right.
print("Hi".ljust(5, "*"))

#rjust()- Places text on the right and fills the remaining space on the left.
print("Hi".rjust(5, "*"))

#zfill()- Adds zeros to the beginning of a string until it reaches the given length.
print("42".zfill(5))
```

```python
#7. Other useful methods
# min()- Finds the character with the smallest Unicode value.
print(min("olacle"))

#max- Finds the character with the largest Unicode value.
print(max("Python"))

#sorted()- Arranges the characters in sorted order and returns a list.
print(sorted("cab"))



```

```python
#PROTECTED
class Rectangle():
  def __init__(self,l,b):  #instead of get data here v will give constructor
    self._length=l
    self.breadth=b
  def display_data(self):
    print(self._length)
    print(self.breadth)
  def area(self):
    print(self._length*self.breadth)
  def perimeter(self):
    print(2*(self._length+self.breadth))
r1=Rectangle(12,15)   #while objecct creation pass the value here it will go to constructor so v added l and b there

r1._length  # can also access the attributes outside by giving _
```

```python
#3. PRIVATE
class Rectangle():
  def __init__(self,l,b):
    self.__length=l         #give two underscores before the attributes
    self.breadth=b
  def display_data(self):
    print(self.__length)
    print(self.breadth)
  def area(self):
    print(self.__length*self.breadth)
  def perimeter(self):
    print(2*(self.__length+self.breadth))
r1=Rectangle(12,15)   #while objecct creation pass the value here it will go to constructor so v added l and b there

r1.__length   #can access inside by adding __ but cannot access outside, ERROR WILL OCCUR
```

```python
#using constructor
class Student():
  def __init__(self,n,r):
    self.name=n
    self.roll_no=r
  def display_data(self):
    print(self.name)
    print(self.roll_no)
class Test(Student):
  def __init__(self,n,r,t):
    Student.__init__(self,n,r)
    self.tmark=t
  def display_tmark(self):
    print(self.tmark)
class Sports():
  def __init__(self,s):
    self.smark=s
  def display_smark(self):
    print(self.smark)
class Grade(Test,Sports):
  def __init__(self,n,r,t,s):       #as this is the child and child's parents are test nd mark
    Sports.__init__(self,s)
    Test.__init__(self,n,r,t)

  def calgrade(self):
    self.tomark = self.tmark + self.smark


    if self.tomark>80:
      print("Grade A")
    elif self.tomark>50:
      print("Grade B")
    else:
      print("failed")

g1=Grade("abu",42,40,20)
#g1.get_smark()
#g1.get_tmark()
g1.calgrade()
```

```python
#operator overloading
class Addition():
  def  __init__(self,n):
    self.value=n      # here the value wll b 10 and in OTHER it will store 12
  def __add__(self,other):             #a1+a2 will be stored here as a1.__add__(a2)
    return self.value+other.value
a1=Addition(10)
a2=Addition(12)
print(a1+a2)
```

```python
#exception handling

try:                           #in what all code error will come put them inside the try block
  d=int(input("Enter a number: "))
  b=10/d

except ValueError:              #copy the error that will come, if its used
  print("denominator must be a digit")

except ZeroDivisionError:
  print("denominator cannot be zero")

else:
  print(b)


finally:                          #error undelum illelum use cheyyunnathan FINALLLY
  print("program finished")
```
